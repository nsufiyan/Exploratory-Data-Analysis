**Method 2 np.where**
- it works only for two label such case_status or full_time_position
- where only two condition
- if case_status=="Certified" then 0
- else 1(DENIED)

```python
file_path="C:\\Users\\user\\Downloads\\datasets\\Visadataset.csv"
visa_df=pd.read_csv(file_path)
visa_df.head()
```

```python
con1=visa_df["case_status"]=="Certified"
visa_df["case_status"]=np.where(con1,0,1) #np.where(<condition>,<if condition true then this>,<if false then this>)
visa_df.head()
```
