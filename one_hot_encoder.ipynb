**One Hot Encoder:**
- one hot encoder means at a time one is on(1/True) other is off(0/False)

|case_Status|certified|denied|
|-----------|---------|------|
|certified  |   1     | 0    |
|Denied     |   0     | 1    |

**advantages**
- one hot encoder new columns are orthogonal each other

- orthogonally means 90 degree phase shift

- 90 degree means perpendicular means independent to each other

- independent means no relation

- it is very important property

**Disadvantages:**
- assume that u have 100 unique labels for a column

- so it will create 100 new columns

- to process 100 columns we require more memory,more time and more hardware

- this is called curse of dimensionality
