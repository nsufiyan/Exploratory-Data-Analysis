-  converting all skewed distribution into Normal Distribution is very important

- beacuse it is assumped that all maths is develope by assumping data follows normal distribution

**Methods**
- log transformation

- square root transformation

- exponential transformation

- box-cox tranformation

- yeo-jhanosn transformation


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=np.random.exponential(size=10000)
plt.hist(data,bins=40,label="Exponential")
plt.legend()
plt.show()
```

**appply log tranformation**

- if there is exponential data like above aplly natural log 
- loga**n=nloga
- removed the skewed but not completly convert into normal distribution
- donot woorked with negative value

```python
log_data=np.log(data) 
plt.hist(log_data,bins=40,label="log") #skewness is removed
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.hist(data,bins=40,label="exponential")
plt.legend()
plt.subplot(1,2,2)
plt.hist(log_data,bins=40,label="log")
plt.legend()
```

**Reciprocol tranformation**
- cannot use if data value is 0

```python
rec_data=1/data 
plt.hist(rec_data,bins=40,label="reciprocal") #skewness is removed
plt.legend()
plt.show()
```

**square root tranformation**

```python
sq_data=np.sqrt(data)
plt.hist(sq_data,bins=40,label="square roott transformation") #skewness is removed
plt.legend()
plt.show()
```
