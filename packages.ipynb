**step 6: how to drop/delete the column**

- what is your dataframe name
- axis=1 =====> columns
- axis=0 =====> rows
- what is name of the columns===>city
- when u drop the column you are modify the table
- then dataframe should be overwrite  ====>inplace means data will overwrite
-  or you want to create a new dataset

```python
names.drop('city',inplace=True,axis=1)   #axis=1 indicates columns 
                                         #inplace=True overwrites    
```
