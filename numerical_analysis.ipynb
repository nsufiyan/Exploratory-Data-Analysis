**Percentile and quantile**

- percentile ranges from 1 to 100
- quantile means
   - q1=25 percentile
   - q2=50 percentile
   - q3=75 percentile

- np.percentile(<direct number bw 1 to 100>,data)
    - np.percentile(75,data)
- np.quantile(<decimal number>,data)
    - np.quantile(0.75,data)

**Both quantile and percentile provides same answer**

```python
q1=round(np.percentile(visa_df["prevailing_wage"],25),2)
q2=round(np.percentile(visa_df["prevailing_wage"],50),2)
q3=round(np.percentile(visa_df["prevailing_wage"],75),2)

q1,q2,q3
```

**how to find empherical rule**

when data follows normal distribution
- u-1*sd to u+1*sd =68%
- u-2*sd to u+2*sd =95%
- u-3*sd to u+3s*d =99.7%

```python
val_minus_1=wage_mean-1*wage_std
val_minus_2=wage_mean-2*wage_std
val_minus_3=wage_mean-3*wage_std
val_plus1=wage_mean+1*wage_std
val_plus2=wage_mean+2*wage_std
val_plus3=wage_mean+3*wage_std
```

```python
val_minus_1 ,val_plus1
##########68%###########

val_minus_2,val_plus2
#########95%#########

val_minus_3,val_plus3
#########99.7%########
```

```python
print(f"68% of the total oberservation has value between{val_minus_1 ,val_plus1} ")
print(f"95% of the total oberservation has value between{val_minus_2,val_plus2} ")
print(f"99.7% of the total oberservation has value between{val_minus_3,val_plus3} ")
```

**retrive 68,95,97.7 percentage and check**

```python
cond1=visa_df["prevailing_wage"]>val_minus_1
cond2=visa_df["prevailing_wage"]<val_plus1
```

```python
len(visa_df[cond1&cond2])
```

```python
len(visa_df[cond1&cond2])/len(visa_df)  #67 we need 68 to follow empherical rule
```

```python
cond1=visa_df["prevailing_wage"]>val_minus_2
cond2=visa_df["prevailing_wage"]<val_plus2
len(visa_df[cond1&cond2])/len(visa_df)
```

```python
cond1=visa_df["prevailing_wage"]>val_minus_3
cond2=visa_df["prevailing_wage"]<val_plus3
len(visa_df[cond1&cond2])/len(visa_df)
```
