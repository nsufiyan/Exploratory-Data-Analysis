**Method- 3**
- methods used to fill missing values
- pad

- bfill

- backfill

- ffill

```python
d1={"Name":["ram","sita","laxman",np.nan],
   "Age":[31,32,np.nan,33],
    "City":[np.nan,"hyd","Pune","Chennai"]
   }
df=pd.DataFrame(d1)
df
```

**method=pad**
- every NAN value is filled with its above value (copied aboved value)

```python
df.fillna(method="pad")
```

**method=bfill**
- in this NAN values is replaced with the below value

```python
d1={"Name":["ram","sita","laxman",np.nan],
   "Age":[31,32,np.nan,33],
    "City":[np.nan,"hyd","Pune","Chennai"]
   }
df=pd.DataFrame(d1)
df
```

```python
df.fillna(method="bfill")
```

**method backfill**
- same as bfill

```python
d1={"Name":["ram","sita","laxman",np.nan],
   "Age":[31,32,np.nan,33],
    "City":[np.nan,"hyd","Pune","Chennai"]
   }
df=pd.DataFrame(d1)
df
```

```python
df.fillna(method="backfill")
```

**method=ffill**
- same as pad 


```python
d1={"Name":["ram","sita","laxman",np.nan],
   "Age":[31,32,np.nan,33],
    "City":[np.nan,"hyd","Pune","Chennai"]
   }
df=pd.DataFrame(d1)
df
```

```python
df.fillna(method="ffill",axis=1) #wrng o/p
```

|**Method -4**
- mean,median,mode
- mean,median is used for numerical datta
- mode is used for categorical data

```python
d1={"Name":["ram","sita","laxman",np.nan],
   "Age":[31,32,np.nan,33],
    "City":[np.nan,"hyd","Pune","Chennai"]
   }
df=pd.DataFrame(d1)
df
```

**mean**
- instead of providing the direct random value to missing value fill that value with mean of that column
- same as median and mode

```python
df["Age"].fillna(df["Age"].mean(),inplace=True) #mean is 32 
```

```python
df
```
