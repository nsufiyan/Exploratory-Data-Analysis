**normization**
- ranges between 0 to 1 only

```python
x_min=visa_df["prevailing_wage"].min()
x_max=visa_df["prevailing_wage"].max()
num=visa_df["prevailing_wage"]-x_min
deno=x_max-x_min
X=num/deno
visa_df["prevailing_wage_norm"]=X
visa_df.head()
```

**STANDARIZATON**

```python
file_path="C:\\Users\\user\\Downloads\\datasets\\Visadataset.csv"
visa_df=pd.read_csv(file_path)
visa_df.head()
```

```python
mean=visa_df["prevailing_wage"].mean()
std=visa_df["prevailing_wage"].std()
```

```python
Z_score=(visa_df["prevailing_wage"]-mean)/std
Z_score.head(3)
```

```python
visa_df["prevailing_wage_z_score"]=Z_score
visa_df.head(2)
```

```python
plt.subplot(1,2,1)
plt.boxplot(visa_df["prevailing_wage"])
plt.subplot(1,2,2)
plt.boxplot(visa_df["prevailing_wage_z_score"])
```
