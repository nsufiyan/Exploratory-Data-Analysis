$iloc - loc$

  -**iloc**
- to get range of columns like 100 to 150 use iloc and loc
- syntax: iloc[rows:rows,columns:columns]
- iloc[start:end,start:end]
- if column position is left empty returns all columns
- for specific rows we can also use iloc rather than take

```python
visa_df.iloc[[5,6,9],[2,4,5]]
```

```python
visa_df.iloc[100:150] # returns 100 to 149 rows with all columns
```

```python
visa_df.iloc[50:55,2:5]
```

- for specific rows we can also use iloc rather than take

```python
visa_df.iloc[[100,65,7]] # visa_df.take([100,65,7]) same result
```

```python
visa_df.iloc[[100,65,7],[2,3,6]] #visa_df.take([100,65,7]).take([2,3,6],axis=1) same result
```

```python
visa_df.take([100,65,7]).take([2,3,6],axis=1)
```
